```python
# Configuración Inicial e Importaciones
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, FloatSlider
import ipywidgets as widgets

def mostrar_imagenes(imagenes, titulos, figsize=(15, 5)):
    """Función auxiliar para mostrar múltiples imágenes lado a lado"""
    n = len(imagenes)
    fig, axs = plt.subplots(1, n, figsize=figsize)
    if n == 1: axs = [axs]
    for i in range(n):
        img = imagenes[i]
        if len(img.shape) == 3:
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        axs[i].imshow(img, cmap='gray', vmin=0, vmax=255)
        axs[i].set_title(titulos[i])
        axs[i].axis('off')
    plt.show()

print("Entorno configurado. Asegúrate de subir todas las imágenes necesarias (.tif, .png, .jpg) a Colab.")
```

### 4.2 Filtro Bilateral vs Pasa-Bajos
El filtro bilateral es excelente para preservar bordes mientras elimina el ruido

```python
# --- Caso 2: Filtro Bilateral vs Filtros Clásicos y Análisis de Perfiles ---
def comparar_bilateral(img_path, fila_analisis):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    if img is None: return f"Carga '{img_path}'"

    # Aplicar un filtro pasa-bajos clásico (ej: Gaussiano)
    img_gauss = cv2.GaussianBlur(img, (9, 9), 0)

    # Aplicar Filtro Bilateral
    # cv2.bilateralFilter(src, d, sigmaColor, sigmaSpace)
    img_bilateral = cv2.bilateralFilter(img, 9, 75, 75)

    # Mostrar imágenes
    mostrar_imagenes([img, img_gauss, img_bilateral],
                     ["Original", "Gaussiano", "Bilateral"])

    # --- Ejercicio 4.3: Extracción de perfiles de grises superpuestos ---
    perfil_orig = img[fila_analisis, :]
    perfil_gauss = img_gauss[fila_analisis, :]
    perfil_bilateral = img_bilateral[fila_analisis, :]

    plt.figure(figsize=(15, 4))
    plt.plot(perfil_orig, label='Original', color='black', alpha=0.5)
    plt.plot(perfil_gauss, label='Gaussiano', color='blue')
    plt.plot(perfil_bilateral, label='Bilateral', color='red')
    plt.title(f"Comparación de Perfiles de Intensidad (Fila {fila_analisis})")
    plt.legend()
    plt.grid(True)
    plt.show()

# Probar con las imágenes solicitadas en la guía: mariposa02.png, flores02.jpg, lapices02.jpg
# Usamos un slider para elegir qué fila analizar visualmente (reemplaza el click del mouse)
interact(comparar_bilateral,
         img_path=['mariposa02.png', 'flores02.jpg', 'lapices02.jpg'],
         fila_analisis=IntSlider(min=0, max=300, step=1, value=150))
```
